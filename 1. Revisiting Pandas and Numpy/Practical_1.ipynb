```python
import  pandas as pd
import numpy as np
df = pd.read_csv("/content/CategoricalDataset.csv")
print(df.head())
```

```python
print(df.info())
#view column datatypes and checking missing value
```

```python
matrix_A = df.iloc[0:240, : ].to_numpy()
# Extract the data into a NumPy array
print(matrix_A)
```

```python
print("Using numpy find out: shape, size, dimension")
print("Shape:", matrix_A.shape)
print("Dimensions:", matrix_A.ndim)
print("Size:", matrix_A.size)
```

```python
print("Matrix Operations")

matrix_B = np.ones((240, 240), dtype=int)
addition = matrix_A + matrix_B
print(addition)
subtraction = matrix_A - matrix_B
print(subtraction)
multiplication = matrix_A * matrix_B
print(multiplication)
division = matrix_A / matrix_B
print(division)
transpose = matrix_A.T
print(transpose)
dot_product = np.dot(matrix_A, matrix_B)
```
