```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# i upload the teaching rating dataset and copy the path of the file and load in csv file using pandas
```

```python
df = pd.read_csv("/content/TeachingRatings.csv")
print(df.head())
```

```python
#the prof column identify the professors i used duplicate to find professors who appear more than one keep=false display all obsevations belonging to duplicate professors.
duplicates = df[df.duplicated(subset='prof', keep=False)]

print("Duplicate cases:")
print(duplicates)

# Number of unique instructors
print("Number of unique instructors:", df['prof'].nunique())

# Mean and standard deviation using all observations
print("Mean age:", df['age'].mean())
print("Standard deviation of age:", df['age'].std())

# Keep only one observation for each professor
no_duplicates_df = df.drop_duplicates(subset='prof')

print("Number of observations after removing duplicates:")
print(len(no_duplicates_df))

# Mean and standard deviation after removing duplicates
print("Mean age after removing duplicates:",
      no_duplicates_df['age'].mean())

print("Standard deviation after removing duplicates:",
      no_duplicates_df['age'].std())
```

```python
#2. Using a bar chart, demonstrate if instructors teaching lower-division courses receive higher average teaching evaluations
division_eval = df.groupby('division')['eval'].mean()

print(division_eval)

# Bar chart
plt.figure(figsize=(7, 5))

division_eval.plot(kind='bar')

plt.title("Average Teaching Evaluation by Course Division")
plt.xlabel("Course Division")
plt.ylabel("Average Evaluation")
plt.xticks(rotation=0)
plt.show()
```

```python
# If the bar for lower division is higher than upper division, then lower-division instructors receive higher average evaluations.
# Otherwise, lower-division instructors do not receive higher average evaluations.
```

```python
#3. Plot the relationship between age and teaching evaluation scores.
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='age',
    y='eval'
)

plt.title("Relationship Between Age and Teaching Evaluation")
plt.xlabel("Age")
plt.ylabel("Teaching Evaluation")
plt.show()
```

```python
#Look at the overall direction of the points. If the points generally move downward as age increases, there is a negative relationship. If they move upward, there is a positive relationship. If the points are widely scattered, the relationship is weak.
```

```python
#4. Using gender-differentiated scatter plots, plot the relationship between age and teaching evaluation scores.
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='age',
    y='eval',
    hue='gender'
)

plt.title("Age vs Teaching Evaluation by Gender")
plt.xlabel("Age")
plt.ylabel("Teaching Evaluation")
plt.legend(title="Gender")
plt.show()
```

```python
# The different groups allow us to see whether the relationship between age and evaluation appears different for male and female instructors.
```

```python
# 5. Create a scatterplot of age and evaluation scores, differentiated by gender and tenure.
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='age',
    y='eval',
    hue='gender',
    style='tenure',
    s=80
)

plt.title("Age vs Teaching Evaluation by Gender and Tenure")
plt.xlabel("Age")
plt.ylabel("Teaching Evaluation")
plt.legend(title="Gender / Tenure")
plt.show()
```
