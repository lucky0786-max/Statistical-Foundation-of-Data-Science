```python
print("I loaded the Student Rating dataset from a CSV file using Pandas. The head() function displays the first five rows so that I can check the data.")
```

```python
df = pd.read_csv("/content/TeachingRatings(1).csv")
df.head()
print(df.columns)
print(df.shape)
print(df.info())
```

```python
print("First, I selected only the professors who are visible minorities. Then I checked how many of these professors are tenured. Finally, I multiplied the result by 100 to get the percentage")
```

```python

print("1.Calculate the percentage of visible minorities are tenure professors. Will you say that tenure status differed if teacher was a visible minority")
minority_df = df[df['minority'] == 'yes']
tenure_minority_pct = (minority_df['tenure'].value_counts() / len(minority_df)) * 100
print(tenure_minority_pct)
```

```python
print("I used cross-tabulation to compare minority status and tenure status. normalize=index converts the counts into percentages for each minority group.")
```

```python
print("grouped the professors according to their tenure status. Then I calculated the mean age and standard deviation for each group.")
```

```python
print("2. Does average age differ by tenure? Produce the means and standard deviations for both tenured and untenured professors")
tenure_df = df[df['tenure'] == 'yes']
notenure_df = df[df['tenure'] == 'no']
print("Mean age of tenured", tenure_df['age'].mean())
print("SD of tenured", tenure_df['age'].std())
print("Mean age of untenured", notenure_df['age'].mean())
print("SD of untenured", notenure_df['age'].std())
```

```python
print("Age is a numerical variable, so a histogram is suitable. The histogram shows how many professors fall into different age ranges.")
```

```python
print("3. Which graph works better and plot that graph for the age variable?")
tenure_df = df[df['tenure'] == 'yes']
notenure_df = df[df['tenure'] == 'no']
plt.hist(tenure_df['age'], bins=10, alpha=0.5, label='tenure')
plt.hist(notenure_df['age'], bins=10, alpha=0.5, label='not tenure')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution by Tenure')
plt.legend()
plt.show()
```

```python
print("Gender is a categorical variable, so I used a bar graph. value_counts() counts the number of professors in each gender category. The bar graph then displays these counts.")
print("pyplot.barh() creates a horizontal bar graph. The gender categories are shown on the Y-axis and the number of professors is shown on the X-axis.")
```

```python
3 print("Define the difference between pyplot.bar and pyplot.barh. Also plot the graph for gender variable")
gender_counts = df['gender'].value_counts()
plt.barh(gender_counts.index, gender_counts.values)
gender_df = df['gender'].value_counts()
plt.bar(gender_df.index, gender_df.values)
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Gender Distribution')
plt.show()
```
