```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import statsmodels.api as sm
import statsmodels.formula.api as smf

from scipy import stats
```

```python
# Load dataset

df = pd.read_csv("/content/TeachingRatings(1).csv")

# Display first five rows
print(df.head())

# Display column names
print(df.columns)

# Check missing values
print(df.isnull().sum())
```

```python
correlation = df["eval"].corr(df["beauty"])
print("Correlation:", correlation)
```

```python
# Regression with T-test
# Remove missing values

data = df[['gender', 'eval']].dropna()

# Apply regression model

model = smf.ols('eval ~ C(gender)', data=data).fit()

# Display regression results

print(model.summary())

# Perform T-test

print("\nT-test result:")
print(model.t_test('C(gender)[T.male] = 0'))
```

We select gender and evaluation columns.
We remove missing values.
We apply linear regression.
We perform a T-test to check whether the gender coefficient is significantly different from zero.

We check the p-value.

Result :
If p-value is less than 0.05, gender has a statistically significant association with teaching evaluation.
If p-value is greater than 0.05, there is insufficient evidence of a significant gender association.

```python
# Regression with ANOVA
# Select required columns

data = df[['age', 'beauty']].dropna()

# Create three age groups

data['age_group'] = pd.cut(
    data['age'],
    bins=[0, 40, 50, float('inf')],
    labels=['Young', 'Middle', 'Older'],
    right=False
)

# Apply ANOVA regression model

model = smf.ols('beauty ~ C(age_group)', data=data).fit()

# Display ANOVA table

anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```

We select age and beauty columns.



We divide instructors into three age groups.



We apply the ANOVA test.



We compare beauty scores across the three groups.



We check the F-value and p-value.

If p-value is less than 0.05, beauty scores differ significantly across at least two age groups.



If p-value is greater than 0.05, there is insufficient evidence of a difference.
